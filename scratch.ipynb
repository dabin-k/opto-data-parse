```python
from pathlib import Path

import h5py
import scipy.io

dir = f"{BASE_DIR}/M150605_ICTP1/1"

files = [
    Path("M150605_ICTP1_s1_V1.mat"),
    Path("M150605_ICTP1_s1_not9.mat"),
    Path("1/1_1_M150605_ICTP1_1_1.mat"),
    Path("1/1_1_M150605_ICTP1_Timeline.mat"),
    Path("1/Protocol.mat"),
]

files = [Path(dir) / f for f in files]

def inspect_hdf5(path: Path) -> None:
    """Inspect MATLAB v7.3/HDF5 files recursively."""
    with h5py.File(path, "r") as f:
        def visitor(name, obj):
            if isinstance(obj, h5py.Dataset):
                print(
                    f"{name:70s} "
                    f"shape={str(obj.shape):20s} dtype={obj.dtype}"
                )

        f.visititems(visitor)


for path in files:
    print(f"\n{'=' * 20} {path} {'=' * 20}")

    try:
        # Works for MATLAB files before v7.3.
        for name, shape, dtype in scipy.io.whosmat(path):
            print(f"{name:50s} shape={str(shape):20s} type={dtype}")

    except (NotImplementedError, ValueError, OSError):
        # MATLAB v7.3 files are HDF5 containers.
        try:
            inspect_hdf5(path)
        except OSError as exc:
            print(f"Could not read {path}: {exc}")
```

## Figure 1C — Raster plots: 6 putative E neurons, 2-ms E pulse

```python
import sys
sys.path.insert(0, '/home/dabin/code/ichun_opto')

import numpy as np
# import matplotlib
# matplotlib.use('Agg')
import matplotlib.pyplot as plt
import h5py
import scipy.io

from data_loader import load_data, _load_protocol_conditions

BASE_DIR = "/mnt/scratch/M150605_ICTP1"
SESSION_DIR = f"{BASE_DIR}/1"

# All TTL pulse experiments (single- and paired-pulse; excludes visual tuning 1,14 and reg-pulse 8,10-12)
PULSE_EXPS = [2, 3, 4, 5, 6, 7, 13]

resp, stim, t, tinfo, uinfo = load_data(
    base_dir=BASE_DIR,
    pre_s=0.1,
    post_s=0.9,
    bin_s=0.001,
    selected_exps=PULSE_EXPS,
)
print(f"responses: {resp.shape}   ({resp.shape[0]} trials × {resp.shape[1]} units × {resp.shape[2]} bins)")
print(f"time axis: {t[0]:.3f} to {t[-1]:.3f} s")
```

```python
# Classify each unit as E (wide spike) or I (narrow) using mean waveform trough-to-peak time.
# The kwx stores per-spike filtered waveforms (n_spikes, 20 samples, 8 channels).
# We average across spikes belonging to each unit, then measure trough→peak on the dominant channel.

SAMPLE_RATE  = 30_000
EI_THRESH_MS = 0.4   # paper default: wide >= 0.4 ms → E

kwik_path = f"{BASE_DIR}/1/M150605_ICTP1_s1_not9.kwik"
kwx_path  = f"{BASE_DIR}/1/M150605_ICTP1_s1_not9.kwx"

n_units   = len(uinfo)
unit_ei   = np.zeros(n_units, dtype=int)    # 0 = E (wide), 1 = I (narrow)
unit_ttp  = np.full(n_units, np.nan)        # trough-to-peak ms per unit

with h5py.File(kwik_path, "r") as fk, h5py.File(kwx_path, "r") as fx:
    for cg_key in sorted(fk["channel_groups"].keys()):
        cg       = int(cg_key)
        cids     = fk[f"channel_groups/{cg_key}/spikes/clusters/main"][()]
        wavs     = fx[f"channel_groups/{cg_key}/waveforms_filtered"][()].astype(np.float32)
        # wavs: (n_spikes, 20 samples, 8 channels)

        for unit in uinfo:
            if unit["channel_group"] != cg:
                continue
            mask = cids == unit["cluster_id"]
            if not mask.any():
                continue

            mean_wav = wavs[mask].mean(axis=0)                          # (20, 8)
            dom_ch   = (mean_wav.max(axis=0) - mean_wav.min(axis=0)).argmax()
            dom      = mean_wav[:, dom_ch]

            trough_i = int(dom.argmin())
            # Peak = first maximum strictly after the trough
            search   = dom.copy()
            search[:trough_i] = -np.inf
            peak_i   = int(search.argmax())

            ttp = (peak_i - trough_i) / (SAMPLE_RATE / 1000.0)
            unit_ttp[unit["unit_idx"]] = ttp
            unit_ei [unit["unit_idx"]] = 0 if ttp >= EI_THRESH_MS else 1

e_units = np.where(unit_ei == 0)[0]
i_units = np.where(unit_ei == 1)[0]
print(f"E (wide,  ttp ≥ {EI_THRESH_MS} ms): {len(e_units)} units")
print(f"I (narrow, ttp <  {EI_THRESH_MS} ms): {len(i_units)} units")

# Sanity check: ttp distribution
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(unit_ttp[~np.isnan(unit_ttp)], bins=40, color='steelblue', edgecolor='white')
ax.axvline(EI_THRESH_MS, color='red', linestyle='--', label=f'{EI_THRESH_MS} ms threshold')
ax.set_xlabel("Trough-to-peak (ms)")
ax.set_ylabel("Unit count")
ax.set_title("Waveform width distribution (E/I classification check)")
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
# Build (exp, cond_id) → condition metadata using Protocol.mat for each experiment.
from pathlib import Path

cond_lookup = {}
for exp in PULSE_EXPS:
    protocol_path = Path(SESSION_DIR) / str(exp) / "Protocol.mat"
    if not protocol_path.exists():
        continue
    try:
        for cid, params in _load_protocol_conditions(protocol_path).items():
            cond_lookup[(exp, cid)] = params
    except (KeyError, ValueError):
        pass

# All single-E trials (for ranking); 2ms-E trials only (for plotting)
e2_trials = [i for i, d in enumerate(tinfo)
             if cond_lookup.get((d["exp"],d["cond_id"]),{}).get("exp_type")=="single_E"
             and cond_lookup.get((d["exp"],d["cond_id"]),{}).get("dur_ms")==2.0]
print(f"2-ms single-E trials: {len(e2_trials)}")

# Per-unit firing rate in three windows (on the 2ms E trials)
t_pre   = np.searchsorted(t, -0.1)
t0      = np.searchsorted(t, 0.0)
t_burst = np.searchsorted(t, 0.020)
t_supp  = np.searchsorted(t, 0.150)

sub = resp[e2_trials]
baseline_hz   = sub[:, :, t_pre:t0].mean(axis=(0, 2))    / 0.001
initial_hz    = sub[:, :, t0:t_burst].mean(axis=(0, 2))  / 0.001
suppressed_hz = sub[:, :, t_burst:t_supp].mean(axis=(0, 2)) / 0.001

init_ratio = initial_hz    / np.maximum(baseline_hz, 0.1)
supp_ratio = suppressed_hz / np.maximum(baseline_hz, 0.1)

# Select units with E-neuron response pattern:
#   - meaningful baseline (≥ 1 Hz) so suppression is interpretable
#   - initial burst  ≥ 1.2× baseline
#   - suppression   < 0.80× baseline
e_like = np.where(
    (baseline_hz >= 1.0) &
    (init_ratio  >= 1.2) &
    (supp_ratio  <  0.80)
)[0]
print(f"Units with baseline≥1Hz + burst→suppression: {len(e_like)}")

score      = init_ratio * (1.0 - supp_ratio)
top6_units = e_like[np.argsort(score[e_like])[::-1][:6]]

print(f"\nTop-6 unit indices: {top6_units}")
for u in top6_units:
    print(f"  unit {u:3d}: baseline={baseline_hz[u]:.1f}Hz  "
          f"init={initial_hz[u]:.1f}Hz ({init_ratio[u]:.1f}×)  "
          f"supp={suppressed_hz[u]:.1f}Hz ({supp_ratio[u]:.2f}×)")
```

```python
# Figure 1C: raster plots for 6 putative E neurons, 2-ms E pulse
colors = ['tab:red', 'tab:orange', 'tab:green', 'tab:blue', 'tab:purple', 'tab:brown']
n_tr   = len(e2_trials)

fig, axes = plt.subplots(1, 6, figsize=(12, 3.5), sharey=True)

for ax, uid, color in zip(axes, top6_units, colors):
    for row, trial_idx in enumerate(e2_trials):
        spike_bins = np.where(resp[trial_idx, uid] > 0)[0]
        ax.scatter(
            t[spike_bins], np.full(len(spike_bins), row + 1),
            c=color, s=6, marker='|', linewidths=0.8,
        )
    ax.axvline(0, color='red', lw=1.0, alpha=0.85)
    ax.set_xlim(-0.05, 0.8)
    ax.set_ylim(0, n_tr + 1)
    ax.set_xlabel("Time (s)", fontsize=8)
    ax.tick_params(labelsize=7)
    ax.spines[["top", "right"]].set_visible(False)

axes[0].set_ylabel("Trial", fontsize=8)
fig.suptitle("Fig. 1C  —  6 putative E neurons, 2-ms E pulse", fontsize=9)
plt.tight_layout()
plt.savefig("fig1c_rasters.png", dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved fig1c_rasters.png  ({n_tr} trials, units: {top6_units})")
```
